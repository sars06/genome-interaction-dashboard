```python

#copy of the original dataset
df_clean = df.copy()

#midpoint of interacting genomic region
df_clean["Interactor_Mid"] = (
    (df_clean["Interactor_Start"] + df_clean["Interactor_End"]) / 2
).astype(int)

#width of the interacting region
df_clean["Interactor_Width"] = (
    df_clean["Interactor_End"] - df_clean["Interactor_Start"]
).abs()

# Cis / Trans interaction
df_clean["Interaction_Type"] = np.where(
    df_clean["Feature_Chr"] == df_clean["Interactor_Chr"],
    "cis",
    "trans"
)

# Use absolute genomic distance as primary distance feature
df_clean["Genomic_Distance"] = df_clean["abs_distance"]

# Log-transformed genomic distance
df_clean["Log_Genomic_Distance"] = np.log10(df_clean["Genomic_Distance"])

print("Feature engineering completed!")

print("\nNew columns created:")
print(df_clean.columns[-5:].tolist())

display(
    df_clean[
        [
            "Feature_Chr",
            "Interactor_Chr",
            "Interactor_Start",
            "Interactor_End",
            "Interactor_Mid",
            "Interactor_Width",
            "Interaction_Type",
            "Genomic_Distance",
            "Log_Genomic_Distance"
        ]
    ].head()
)
```

```python


# Define distance ranges in base pairs
distance_bins = [
    0,
    100_000,
    500_000,
    1_000_000,
    5_000_000,
    10_000_000,
    np.inf
]

distance_labels = [
    "0-100 kb",
    "100-500 kb",
    "500 kb-1 Mb",
    "1-5 Mb",
    "5-10 Mb",
    ">10 Mb"
]

# Create distance categories
df_clean["Distance_Category"] = pd.cut(
    df_clean["Genomic_Distance"],
    bins=distance_bins,
    labels=distance_labels,
    include_lowest=True
)

print("Distance categories created!")

print("\nNumber of interactions in each category:")
display(
    df_clean["Distance_Category"]
    .value_counts()
    .sort_index()
)
```

```python
#Treatment-Specific Interaction Strength


# Average the two replicates for each treatment

df_clean["Normal_Strength"] = df_clean[
    ["MN1_SuppPairs", "MN2_SuppPairs"]
].mean(axis=1)

df_clean["Carboplatin_Strength"] = df_clean[
    ["MC1_SuppPairs", "MC2_SuppPairs"]
].mean(axis=1)

df_clean["Gemcitabine_Strength"] = df_clean[
    ["MG1_SuppPairs", "MG2_SuppPairs"]
].mean(axis=1)

# Display the new features
strength_columns = [
    "Normal_Strength",
    "Carboplatin_Strength",
    "Gemcitabine_Strength"
]

print("Treatment-specific interaction strength features created.")

display(
    df_clean[strength_columns].describe().T.round(2)
)
```

```python
#  Treatment Strength Differences


# Calculate pairwise differences between treatment strengths

df_clean["Gemcitabine_vs_Normal"] = (
    df_clean["Gemcitabine_Strength"] -
    df_clean["Normal_Strength"]
)

df_clean["Carboplatin_vs_Normal"] = (
    df_clean["Carboplatin_Strength"] -
    df_clean["Normal_Strength"]
)

df_clean["Gemcitabine_vs_Carboplatin"] = (
    df_clean["Gemcitabine_Strength"] -
    df_clean["Carboplatin_Strength"]
)

# Summarize the differences

difference_columns = [
    "Gemcitabine_vs_Normal",
    "Carboplatin_vs_Normal",
    "Gemcitabine_vs_Carboplatin"
]

difference_summary = (
    df_clean[difference_columns]
    .describe()
    .T
    .round(2)
)

display(difference_summary)
```

```python
#  ML Preprocessing Pipeline

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Categorical features
categorical_features = [
    "Feature_Chr",
    "Strand",
    "Interactor_Chr",
    "Interaction_Type"
]

# Numerical features
numerical_features = [
    "Feature_Start",
    "Annotation",
    "Interactor_Start",
    "Interactor_End",
    "Interactor_Mid",
    "Interactor_Width",
    "Genomic_Distance",
    "Log_Genomic_Distance",
    "Normal",
    "CarboplatinTreated",
    "GemcitabineTreated"
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the already-fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed successfully.")

print("\nOriginal training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
#Baseline Multi-Output Regression

from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create baseline model
baseline_model = DummyRegressor(strategy="mean")

# Train baseline
baseline_model.fit(X_train_processed, y_train)

# Predictions
y_pred_baseline = baseline_model.predict(X_test_processed)

# Calculate metrics
baseline_mae = mean_absolute_error(y_test, y_pred_baseline)
baseline_rmse = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
baseline_r2 = r2_score(y_test, y_pred_baseline)

print("Baseline Model Performance")
print("=" * 40)
print(f"MAE  : {baseline_mae:.4f}")
print(f"RMSE : {baseline_rmse:.4f}")
print(f"R²   : {baseline_r2:.4f}")
```

```python
# Random Forest Multi-Output Regression


from sklearn.ensemble import RandomForestRegressor

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

print("Training Random Forest...")
print("This may take some time for 152,844 rows.")

# Train model
rf_model.fit(X_train_processed, y_train)

print("Random Forest training completed.")

# Predictions
y_pred_rf = rf_model.predict(X_test_processed)

# Overall metrics
rf_mae = mean_absolute_error(y_test, y_pred_rf)
rf_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rf))
rf_r2 = r2_score(y_test, y_pred_rf)

print("\nRandom Forest Performance")
print("=" * 40)
print(f"MAE  : {rf_mae:.4f}")
print(f"RMSE : {rf_rmse:.4f}")
print(f"R²   : {rf_r2:.4f}")
```

```python
# Extra Trees Multi-Output Regression


from sklearn.ensemble import ExtraTreesRegressor

# Create Extra Trees model
extra_model = ExtraTreesRegressor(
    n_estimators=100,
    max_depth=20,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

print("Training Extra Trees...")
print("This may take some time for 152,844 rows.")

# Train model
extra_model.fit(X_train_processed, y_train)

print("Extra Trees training completed.")

# Predictions
y_pred_extra = extra_model.predict(X_test_processed)

# Overall metrics
extra_mae = mean_absolute_error(y_test, y_pred_extra)
extra_rmse = np.sqrt(mean_squared_error(y_test, y_pred_extra))
extra_r2 = r2_score(y_test, y_pred_extra)

print("\nExtra Trees Performance")
print("=" * 40)
print(f"MAE  : {extra_mae:.4f}")
print(f"RMSE : {extra_rmse:.4f}")
print(f"R²   : {extra_r2:.4f}")
```

```python
#  Random Forest Performance: CIS vs TRANS

print("Random Forest Performance by Interaction Type")
print("=" * 60)

# Get processed feature names
processed_feature_names = preprocessor.get_feature_names_out()

# Find the encoded Interaction_Type columns
interaction_columns = [
    (i, name)
    for i, name in enumerate(processed_feature_names)
    if 'Interaction_Type_' in name
]

print("\nEncoded interaction columns:")
print(interaction_columns)

# Find the column representing TRANS
trans_column = [
    i for i, name in interaction_columns
    if name.endswith('_trans')
][0]

# Convert processed test data to numpy array
X_test_array = np.asarray(X_test_processed)

# Identify TRANS rows
trans_mask = X_test_array[:, trans_column] == 1

# CIS is everything that is not TRANS
cis_mask = ~trans_mask

# Convert targets/predictions to numpy arrays
y_test_array = np.asarray(y_test)
y_pred_array = np.asarray(y_pred_rf)

# ------------------------------------------------------------
# CIS performance
# ------------------------------------------------------------

print("\nCIS interactions")
print("-" * 40)
print(f"Number of samples: {cis_mask.sum()}")

cis_actual = y_test_array[cis_mask]
cis_predicted = y_pred_array[cis_mask]

cis_mae = mean_absolute_error(cis_actual, cis_predicted)
cis_rmse = np.sqrt(mean_squared_error(cis_actual, cis_predicted))
cis_r2 = r2_score(cis_actual, cis_predicted)

print(f"MAE  : {cis_mae:.4f}")
print(f"RMSE : {cis_rmse:.4f}")
print(f"R²   : {cis_r2:.4f}")

# ------------------------------------------------------------
# TRANS performance
# ------------------------------------------------------------

print("\nTRANS interactions")
print("-" * 40)
print(f"Number of samples: {trans_mask.sum()}")

trans_actual = y_test_array[trans_mask]
trans_predicted = y_pred_array[trans_mask]

trans_mae = mean_absolute_error(trans_actual, trans_predicted)
trans_rmse = np.sqrt(mean_squared_error(trans_actual, trans_predicted))

# R² cannot be calculated reliably with extremely few samples
if trans_mask.sum() >= 2:
    trans_r2 = r2_score(trans_actual, trans_predicted)
else:
    trans_r2 = np.nan

print(f"MAE  : {trans_mae:.4f}")
print(f"RMSE : {trans_rmse:.4f}")
print(f"R²   : {trans_r2:.4f}")
```
